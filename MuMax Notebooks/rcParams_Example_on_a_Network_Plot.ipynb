```python
"""This script constructs a directed Graphviz diagram mapping magnetic principles to acoustic devices.
"""

import graphviz
import matplotlib.pyplot as plt
from IPython.display import display, Image

# Establish global notebook environment resolution and font defaults
plt.rcParams['figure.dpi'] = 250
plt.rcParams['font.sans-serif'] = ['DejaVu Sans']
plt.rcParams['font.weight'] = 'normal'

# Control Knobs
GRAPH_DPI = str(plt.rcParams['figure.dpi'])
NODE_FONT = "DejaVu Sans"
NODE_FONT_WEIGHT = "normal"
EDGE_ROUTING = "spline"
NODE_SHAPE = "box"
BACKGROUND_COLOR = "white"
FOREGROUND_COLOR = "black"
RANK_DIRECTION = "TB"

if __name__ == '__main__':
    dag = graphviz.Digraph(name='Magnetic_Acoustic_Relations', format='png')

    dag.attr(
        dpi=GRAPH_DPI,
        splines=EDGE_ROUTING,
        bgcolor=BACKGROUND_COLOR,
        rankdir=RANK_DIRECTION
    )
    dag.attr(
        'node',
        shape=NODE_SHAPE,
        style='solid',
        color=FOREGROUND_COLOR,
        fontcolor=FOREGROUND_COLOR,
        fontname=NODE_FONT,
        fontsize='10'
    )
    dag.attr(
        'edge',
        color=FOREGROUND_COLOR,
        fontcolor=FOREGROUND_COLOR,
        arrowhead='normal',
        fontname=NODE_FONT,
        fontsize='9'
    )

    # Define nodes
    dag.node('Magnets', 'Permanent Magnets')
    dag.node('Fields', 'Magnetic Fields')
    dag.node('Speakers', 'Microspeakers')

    # Define labeled edges
    dag.edge('Magnets', 'Fields', label=' generate')
    dag.edge('Fields', 'Speakers', label=' induce sound')

    # Render and display in-memory
    binary_png = dag.pipe(format='png')
    display(Image(data=binary_png))
```
